```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/California.zip')
```

```python
df['total_bedrooms']=df['total_bedrooms'].fillna(df['total_bedrooms'].mean())
```

```python
df = pd.get_dummies(df,columns=['ocean_proximity'],drop_first=True)
```

```python
df.head()
```

```python
x=df.drop('median_house_value',axis=1)
y=df['median_house_value']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
SS=StandardScaler()
```

```python
x_train=SS.fit_transform(x_train)
x_test=SS.transform(x_test)
```

```python
# Linear Regression
from sklearn.linear_model import LinearRegression
LR=LinearRegression()
LR.fit(x_train,y_train)
```

```python
y_pred_lr=LR.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error,root_mean_squared_error,mean_absolute_error,r2_score
```

```python
LR_MAE = mean_absolute_error(y_test, y_pred_lr)
LR_MSE = mean_squared_error(y_test, y_pred_lr)
LR_RMSE = root_mean_squared_error(y_test, y_pred_lr)
LR_R2 = r2_score(y_test, y_pred_lr)
```

```python
print("Linear Regression Values")
print("MAE:", LR_MAE)
print("MSE:", LR_MSE)
print("RMSE:", LR_RMSE)
print("R2:", LR_R2)
```

```python
LR.coef_
```

```python
LR.intercept_
```

```python
# 1. Lasso Regression
from sklearn.linear_model import Lasso

L1=Lasso(alpha=1)
L1.fit(x_train,y_train)

y_pred_lasso=L1.predict(x_test)

Lasso_MAE = mean_absolute_error(y_test, y_pred_lasso)
Lasso_MSE = mean_squared_error(y_test, y_pred_lasso)
Lasso_RMSE = root_mean_squared_error(y_test, y_pred_lasso)
Lasso_R2 = r2_score(y_test, y_pred_lasso)

print("Lasso Regression Values")
print("MAE:", Lasso_MAE)
print("MSE:", Lasso_MSE)
print("RMSE:", Lasso_RMSE)
print("R2:", Lasso_R2)
```

```python
L1.coef_
```

```python
L1.intercept_
```

```python
# 2. Ridge Regression
from sklearn.linear_model import Ridge
L2=Ridge(alpha=1)
L2.fit(x_train,y_train)
y_pred_ridge=L2.predict(x_test)
Ridge_MAE = mean_absolute_error(y_test, y_pred_ridge)
Ridge_MSE = mean_squared_error(y_test, y_pred_ridge)
Ridge_RMSE = root_mean_squared_error(y_test, y_pred_ridge)
Ridge_R2 = r2_score(y_test, y_pred_ridge)
print("Ridge Regression")
print("MAE:", Ridge_MAE)
print("MSE:", Ridge_MSE)
print("RMSE:", Ridge_RMSE)
print("R2:", Ridge_R2)
```

```python
L2.coef_
```

```python
L2.intercept_
```

1. To choose a best fit model:MAE,MSE,RMSE should be low and R2 score should be higher.
2. Among the three models, Ridge Regression performed best on the test data. It produced the lowest MAE (50,699.23), MSE (4.904 billion), and RMSE (70,028.48), while achieving the highest R² score of 0.6258. Lasso and Linear Regression produced almost identical results, but Ridge performed slightly better across all evaluation metrics.
3. Ridge regularization shrinks the model coefficients, which can help reduce overfitting. Therefore, based on the test-set results, Ridge Regression generalizes best among the three models.
